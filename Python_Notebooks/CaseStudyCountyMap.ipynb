```python
import geopandas as gpd
#mount the drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)
#Import county boundary polygons file
file = '/content/drive/My Drive/ESIIL2025MSU/Project Data/Colorado_County_Boundaries.geojson'
#define the variable
counties_gdf = gpd.read_file(file)
#select relevent columns
counties_gdf = counties_gdf[['COUNTY','geometry']]
display(counties_gdf.head())
```

```python
import folium

# Create a base map centered on Colorado
m = folium.Map(location=[39.5, -105.5], zoom_start=7)

# Filter the GeoDataFrame for the specified counties
highlight_counties = ['LOGAN', 'DOUGLAS', 'EAGLE']
highlight_gdf = counties_gdf[counties_gdf['COUNTY'].isin(highlight_counties)]

# Add all county borders to the map
folium.GeoJson(
    counties_gdf,
    name='All County Borders',
    style_function=lambda feature: {
        'fillColor': 'none', # No fill for the base county borders
        'color': 'gray',    # Color for the borders
        'weight': 2,        # Border thickness
        'fillOpacity': 0    # No fill opacity
    }
).add_to(m)


# Add the highlighted counties to the map
folium.GeoJson(
    highlight_gdf,
    name='Highlighted Counties',
    style_function=lambda feature: {
        'fillColor': 'green',
        'color': 'black',
        'weight': 2,
        'fillOpacity': 0.5
    }
).add_to(m)

# Display the map
display(m)
# Save and download the percent change map
m.save('casestudy_highlightcounties.html')
```
