```python
import pandas as pd
```

```python
file_path = '/content/drive/My Drive/ESIIL Project Data/Population Data/Pop Data - Douglas_Eagle_Logan.csv'
df_new = pd.read_csv(file_path)

display(df_new.head())
display(df_new.info())
```

## Data preparation

### Subtask:
Reshape the data to be suitable for ANOVA, which typically requires a long format with columns for population, county, and year.

**Reasoning**:
Clean the population columns and reshape the DataFrame to a long format suitable for ANOVA.

```python
for col in ['Douglas Population', 'Eagle Population', 'Logan Population']:
    df_new[col] = df_new[col].str.replace(',', '').astype(int)

df_melted_new = df_new.melt(id_vars=['Year'],
                            value_vars=['Douglas Population', 'Eagle Population', 'Logan Population'],
                            var_name='County',
                            value_name='Population')

df_melted_new['County'] = df_melted_new['County'].str.replace(' Population', '')

display(df_melted_new.head())
```

```python
print(df_melted_new)
```

## Anova analysis (1985-2001)

### Subtask:
Perform ANOVA on the data for the years 1985 to 2001.

**Reasoning**:
Filter the dataframe for the specified years and perform the ANOVA analysis.

```python
df_1985_2001_new = df_melted_new[(df_melted_new['Year'] >= 1985) & (df_melted_new['Year'] <= 2001)]

from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

model_1985_2001_new = ols('Population ~ C(County)', data=df_1985_2001_new).fit()
anova_table_1985_2001_new = anova_lm(model_1985_2001_new)

print(anova_table_1985_2001_new)
```

## Anova analysis (2002-2022)

### Subtask:
Perform ANOVA on the data for the years 2002 to 2022.

**Reasoning**:
Filter the data for the years 2002-2022, fit an OLS model, perform ANOVA, and print the results.

```python
df_2002_2022_new = df_melted_new[(df_melted_new['Year'] >= 2002) & (df_melted_new['Year'] <= 2022)]

model_2002_2022_new = ols('Population ~ C(County)', data=df_2002_2022_new).fit()
anova_table_2002_2022_new = anova_lm(model_2002_2022_new)

print(anova_table_2002_2022_new)
```

## Anova analysis (1985-2022)

### Subtask:
Perform ANOVA on the data for all years 1985 to 2022.

**Reasoning**:
Perform ANOVA on the entire dataset (1985-2022) using the previously melted and cleaned DataFrame `df_melted_new`.

```python
model_1985_2022_new = ols('Population ~ C(County)', data=df_melted_new).fit()
anova_table_1985_2022_new = anova_lm(model_1985_2022_new)

print(anova_table_1985_2022_new)
```

## Present results

### Subtask:
Display the results of the ANOVA analyses.

**Reasoning**:
Display the results of the three ANOVA analyses by printing the respective ANOVA tables.

```python
print("ANOVA Table for 1985-2001:")
print(anova_table_1985_2001_new)
print("\nANOVA Table for 2002-2022:")
print(anova_table_2002_2022_new)
print("\nANOVA Table for 1985-2022:")
print(anova_table_1985_2022_new)
```

## Export ANOVA Results to CSV

### Subtask:
Export the ANOVA tables for each time period to separate CSV files.

**Reasoning**:
Use the `to_csv()` method to save each ANOVA table DataFrame to a CSV file with an appropriate filename.

```python
# Export ANOVA table for 1985-2001
anova_table_1985_2001_new.to_csv('/content/drive/My Drive/ANOVA_Douglas_Eagle_Logan_1985_2001.csv')

# Export ANOVA table for 2002-2022
anova_table_2002_2022_new.to_csv('/content/drive/My Drive/ANOVA_Douglas_Eagle_Logan_2002_2022.csv')

# Export ANOVA table for 1985-2022
anova_table_1985_2022_new.to_csv('/content/drive/My Drive/ANOVA_Douglas_Eagle_Logan_1985_2022.csv')

print("ANOVA results for Douglas, Eagle, and Logan counties exported to CSV files in your Google Drive.")
```
